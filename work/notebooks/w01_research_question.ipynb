## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**The question:** with way too many pages to check by hand, which ones should someone
actually look at first?

**Unit of analysis:** one page (`content_id`).

**What decision this helps with:** right now there's no real system for this a reviewer
would just be guessing or going by memory. This turns it into an actual ranked list based
on evidence instead of gut feeling.

**Who acts on it:** a content reviewer, basically. They'd start from the top of the list
and go down refresh, expand, or skip depending on the reason attached to each page.

**What a wrong call costs:**
- If I flag something that didn't need fixing, that's wasted time and since review time
  is limited, it means some page that actually needed help got skipped instead.
- If I miss a page that's genuinely dropping off, it just keeps quietly losing traffic
  until someone happens to notice, way later than they should have.

**Why not just a checklist:** because "worth refreshing" isn't one clean number — it's a
mix of things (traffic, position, how old the content is, how thin it is) and none of them
alone tell the full story. That's exactly the kind of thing a model is actually useful for.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os

# Am cloning my repo
if not os.path.exists("flyrank-ml-internship"):
    !git clone https://github.com/muhammadahannadeembaig/flyrank-ml-internship.git

%cd flyrank-ml-internship

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total rows in starter dataset:", len(df))
print()

# 1. Pages that are old/stale but still getting real visibility
stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print(f"Stale but still visible pages: {len(stale_visible)} "
      f"({len(stale_visible)/len(df)*100:.1f}% of all pages)")

# 2. Pages that are declining even though there's still real search demand
declining_with_demand = df[(df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)]
print(f"Declining pages with real demand: {len(declining_with_demand)} "
      f"({len(declining_with_demand)/len(df)*100:.1f}% of all pages)")

# 3. Pages that are thin content but still pulling in traffic
thin_visible = df[(df["word_count"] > 0) & (df["word_count"] < 1200) & (df["impressions_90d"] >= 250)]
print(f"Thin content that's still visible: {len(thin_visible)} "
      f"({len(thin_visible)/len(df)*100:.1f}% of all pages)")
```

Ran this on the 30,000-row starter set and the numbers are pretty uneven:

- **13,152 pages (43.8%)** are marked "declining" while still pulling real search demand
  — that's a huge chunk, way more than I expected.
- Only **17 pages (0.1%)** are stale-but-visible, and **82 (0.3%)** are thin-but-visible —
  both way smaller than the declining group.

So the declining-with-demand signal is clearly the strongest one here, at least on this
slice. The other two might matter more once I'm working with the full warehouse data,
which has way more pages and more history to check freshness against. For now this is
enough evidence that Lane 2 has something real to work with.
